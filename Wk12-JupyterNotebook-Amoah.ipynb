## Unsupervised methods (Weeks 10-11): does clustering recover the same five regions?

| Method | Result |
|---|---|
| K-Means | Silhouette-optimal k=3 (0.386), not k=5 (0.347). ARI highest at k=2 (0.585). |
| DBSCAN | eps=0.6 gives 7 clusters, 2.2% noise overall; noise concentrated in Northeast (22.3%). |
| HAC | Top dendrogram split is two-way; one branch splits again, echoing k=3. Ward best silhouette (0.307); average best ARI (0.263). |

Unlike the supervised methods, none of the unsupervised methods naturally recover a clean five-way partition matching the true regions. This is the central cross-method finding of Milestone 2: strong supervised separability does not imply that unsupervised, label-free geometry organizes into the same groups.

## Cross-method synthesis

1. **Classification is solved; clustering is not trivial.** All six supervised methods tested achieve near-perfect accuracy, but K-means and HAC both independently point to a natural cluster count of two or three, not five.
2. **Class imbalance affects each method differently.** Supervised models lose recall on the smallest class (North). K-means fragments the largest class (Southeast) into multiple clusters. DBSCAN's noise tracks each region's internal geographic homogeneity (Northeast) rather than its size.
3. **Internal validation metrics can mislead.** K-means's silhouette-optimal k and HAC's silhouette-best linkage both diverge from what best matches the true region labels (ARI), a caution for any unsupervised analysis run without ground truth to check against.
